```python
import numpy as np
```

```python
#Logistic function turns values into a probability and values between 0 and 1
def logisticF(sum):
  return (1)/(1+(np.e**(-sum)))
```

```python
#sum takes the inputs of the previous 2 neuros multiplied by the edges and finally adds the bias
# Node h1 weighted sum(1,1)
h1_sum = ((1*4.83) + (1 * (-4.83)))-2.82

#Logistic function h2 output

h1_output = logisticF(h1_sum)
print(h1_output)

# Node h2 weighted sum(1,1)
h2_sum = ((1*4.6) + (1 * (-4.63)))-2.74

#Logistic function h2 output

h2_output = logisticF(h2_sum)
print(h2_output)


# Node o1 weighted sum(1,1)
o1_sum = ((h1_sum*5.73) + (h2_output * (5.83)))-2.86

#Logistic function 01 output

o1_output = logisticF(o1_sum)
print(o1_output)

```
